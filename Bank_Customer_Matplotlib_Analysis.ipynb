Chart 9 — Monthly Transaction Trend → Line Chart

```python
# 9. Monthly Transaction Trend - Line Chart

import pandas as pd
import matplotlib.pyplot as plt

# Load transaction data
transactions = pd.read_csv("data/transactions.csv")

# Convert Transaction_Date to datetime
transactions["Transaction_Date"] = pd.to_datetime(
    transactions["Transaction_Date"]
)

# Monthly transaction amount
monthly_transactions = (
    transactions
    .groupby(transactions["Transaction_Date"].dt.to_period("M"))["Amount"]
    .sum()
)

# Convert period to string for plotting
monthly_transactions.index = monthly_transactions.index.astype(str)

# Create line chart
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_transactions.index,
    monthly_transactions.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Transaction Trend")
plt.xlabel("Month")
plt.ylabel("Total Transaction Amount")

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save chart
plt.savefig(
    "output/09_monthly_transaction_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```

Chart 10 — City-wise Transaction Amount → Horizontal Bar Chart

```python
# 10. City-wise Transaction Amount - Horizontal Bar Chart

import pandas as pd
import matplotlib.pyplot as plt

# Load data
customers = pd.read_csv("data/customers.csv")
accounts = pd.read_csv("data/accounts.csv")
transactions = pd.read_csv("data/transactions.csv")

# Join customer → account → transaction
merged_data = (
    customers
    .merge(accounts, on="Customer_ID", how="inner")
    .merge(transactions, on="Account_ID", how="inner")
)

# Calculate total transaction amount by city
city_transactions = (
    merged_data
    .groupby("City")["Amount"]
    .sum()
    .sort_values(ascending=True)
)

# Create horizontal bar chart
plt.figure(figsize=(10, 6))

plt.barh(
    city_transactions.index,
    city_transactions.values
)

plt.title("City-wise Total Transaction Amount")
plt.xlabel("Total Transaction Amount")
plt.ylabel("City")

plt.tight_layout()

# Save chart
plt.savefig(
    "output/10_city_wise_transaction_amount.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```

Chart 12 — Transaction Amount vs Customer Income → Line Chart

```python
# 12. Transaction Amount vs Customer Income - Line Chart

import pandas as pd
import matplotlib.pyplot as plt

# Load data
customers = pd.read_csv("data/customers.csv")
accounts = pd.read_csv("data/accounts.csv")
transactions = pd.read_csv("data/transactions.csv")

# Merge customer → account → transaction
income_transactions = (
    customers
    .merge(accounts, on="Customer_ID", how="inner")
    .merge(transactions, on="Account_ID", how="inner")
)

# Create income groups
income_transactions["Income_Group"] = pd.cut(
    income_transactions["Annual_Income"],
    bins=5
)

# Calculate average transaction amount for each income group
income_analysis = (
    income_transactions
    .groupby("Income_Group", observed=True)["Amount"]
    .mean()
)

# Create line chart
plt.figure(figsize=(12, 6))

plt.plot(
    income_analysis.index.astype(str),
    income_analysis.values,
    marker="o",
    linewidth=2
)

plt.title("Average Transaction Amount by Income Group")
plt.xlabel("Annual Income Group")
plt.ylabel("Average Transaction Amount")

plt.xticks(rotation=30)
plt.grid(True, alpha=0.3)

plt.tight_layout()

# Save chart
plt.savefig(
    "output/12_transaction_amount_vs_income_line.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```

Chart 13 — Top 10 Customers by Transaction Amount → Horizontal Bar

```python
# 13. Top 10 Customers by Transaction Amount - Horizontal Bar

import pandas as pd
import matplotlib.pyplot as plt

# Load data
customers = pd.read_csv("data/customers.csv")
accounts = pd.read_csv("data/accounts.csv")
transactions = pd.read_csv("data/transactions.csv")

# Merge customer → account → transaction
customer_transactions = (
    customers
    .merge(accounts, on="Customer_ID", how="inner")
    .merge(transactions, on="Account_ID", how="inner")
)

# Calculate total transaction amount by customer
top_customers = (
    customer_transactions
    .groupby(
        ["Customer_ID", "First_Name", "Last_Name"]
    )["Amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .sort_values(ascending=True)
)

# Create customer names
customer_names = [
    f"{first} {last}"
    for (_, first, last) in top_customers.index
]

# Create horizontal bar chart
plt.figure(figsize=(10, 6))

plt.barh(
    customer_names,
    top_customers.values
)

plt.title("Top 10 Customers by Total Transaction Amount")
plt.xlabel("Total Transaction Amount")
plt.ylabel("Customer")

plt.tight_layout()

# Save chart
plt.savefig(
    "output/13_top_10_customers_transaction_amount.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```

Chart 15 — Correlation Analysis → Heatmap

```python
# 15. Correlation Analysis - Heatmap

import pandas as pd
import matplotlib.pyplot as plt

# Load data
customers = pd.read_csv("data/customers.csv")
accounts = pd.read_csv("data/accounts.csv")
transactions = pd.read_csv("data/transactions.csv")

# Select numeric customer columns
customer_numeric = customers[
    ["Age", "Annual_Income"]
].copy()

# Select numeric account columns
account_numeric = accounts[
    ["Balance"]
].copy()

# Combine numeric data
numeric_data = pd.concat(
    [customer_numeric, account_numeric],
    axis=1
)

# Calculate correlation
correlation = numeric_data.corr()

# Create heatmap-style visualization
plt.figure(figsize=(8, 6))

plt.imshow(
    correlation,
    cmap="coolwarm",
    aspect="auto"
)

# Add color bar
plt.colorbar(label="Correlation")

# Add labels
plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=30
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

# Display correlation values
for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(
            j,
            i,
            f"{correlation.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.title("Correlation Analysis")
plt.tight_layout()

# Save chart
plt.savefig(
    "output/15_correlation_analysis.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```

Final Chart 3 — Monthly Transaction Trend → Line Chart

```python
# FINAL 3. Monthly Transaction Trend - Line Chart

import pandas as pd
import matplotlib.pyplot as plt

transactions = pd.read_csv("data/transactions.csv")

# Convert date
transactions["Transaction_Date"] = pd.to_datetime(
    transactions["Transaction_Date"]
)

# Monthly total transaction amount
monthly_data = (
    transactions
    .groupby(
        transactions["Transaction_Date"].dt.to_period("M")
    )["Amount"]
    .sum()
)

# Convert month labels to string
months = monthly_data.index.astype(str)

plt.figure(figsize=(12, 6))

plt.plot(
    months,
    monthly_data.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Transaction Trend")
plt.xlabel("Month")
plt.ylabel("Total Transaction Amount")

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    "output/final_03_monthly_transaction_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```
